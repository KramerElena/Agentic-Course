```python
# Lab 1 — Local Ollama
# Original additions and modifications © 2026 Dr Elena Kramer, Braude College, Israel.
# Adapted course content and third-party materials remain with their respective rights holders.

import os
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
```

## Set up Ollama

This notebook uses `gpt-oss:120b`, which is already installed on the configured Ollama server. If it is removed later, run `ollama pull gpt-oss:120b` on that server.

Set `OLLAMA_BASE_URL` in the project-root `.env` file. The context length is configured on the Ollama server; this notebook does not set it per request.

Ollama exposes an OpenAI-compatible `/v1` endpoint. The OpenAI Python SDK sends the chat requests to the URL configured in `OLLAMA_BASE_URL`.

```python
MODEL = "gpt-oss:120b"
OLLAMA_BASE_URL = os.getenv("OLLAMA_BASE_URL")
if not OLLAMA_BASE_URL:
    raise ValueError("Set OLLAMA_BASE_URL in the project-root .env file")

openai = OpenAI(base_url=OLLAMA_BASE_URL, api_key="ollama")
```

```python
response = openai.models.list()
available_models = [model.id for model in response.data]
print("Available Ollama models:", available_models)
if MODEL not in available_models:
    raise ValueError(f"Model {MODEL!r} is not available on the Ollama server")
```

```python
messages = [{"role": "user", "content": "Tell me a fun fact"}]
```

```python
messages
```

```python
response = openai.chat.completions.create(model=MODEL, messages=messages)
print(response.choices[0].message.content)
```

```python
question_prompt = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question_prompt}]
response = openai.chat.completions.create(model=MODEL, messages=messages)
question = response.choices[0].message.content
print(question)
```

```python

messages = [{"role": "user", "content": question}]
messages
```

```python
response = openai.chat.completions.create(model=MODEL, messages=messages)
answer = response.choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
message = f"""
Here is a question:
{question}

And here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""
display(Markdown(message))
# print(message)
```

```python
messages = [{"role": "user", "content": message}]
response = openai.chat.completions.create(model=MODEL, messages=messages)
print(response.choices[0].message.content)
```
