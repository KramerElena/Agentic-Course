## Cloud setup: OpenRouter

Create an OpenRouter API key and add it to the project-root `.env` file as `OPENROUTER_API_KEY=your_key`. Keep the key private and do not commit it.

This lab uses OpenAI's GPT-4o mini through OpenRouter. API requests are billed by OpenRouter, so check your account's credits and usage limits.

```python
# Lab 1 — OpenRouter Cloud
# Original additions and modifications © 2026 Dr Elena Kramer, Braude College, Israel.
# Adapted course content and third-party materials remain with their respective rights holders.

import os

from dotenv import load_dotenv
from openai import OpenAI

load_dotenv()
```

```python
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
if not OPENROUTER_API_KEY:
    raise ValueError("Add OPENROUTER_API_KEY to the project-root .env file")

OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
MODEL = "openai/gpt-4o-mini"
openai = OpenAI(base_url=OPENROUTER_BASE_URL, api_key=OPENROUTER_API_KEY)
```

The OpenAI Python package is used as a compatible HTTP client, but `base_url` points to OpenRouter. These requests use the cloud model and do not go to Ollama.

```python
messages = [{"role": "user", "content": "Tell me a fun fact"}]
messages
```

```python
response = openai.chat.completions.create(model=MODEL, messages=messages)
print(response.choices[0].message.content)
```

## Generate a question, answer it, and evaluate the answer

First, ask the model to propose a challenging reasoning question.

```python
question_prompt = "Please propose a hard, challenging question to assess someone's IQ. Respond only with the question."
messages = [{"role": "user", "content": question_prompt}]
response = openai.chat.completions.create(model=MODEL, messages=messages)
question = response.choices[0].message.content
print(question)
```

```python
messages = [{"role": "user", "content": question}]
messages
```

```python
answer = openai.chat.completions.create(model=MODEL, messages=messages).choices[0].message.content
print(answer)
```

```python
from IPython.display import Markdown, display

display(Markdown(answer))
```

```python
evaluation_prompt = f"""
Here is a question:
{question}

Here is a possible answer that might be correct or incorrect:
{answer}

Please evaluate if the answer is correct or incorrect.
"""
print(evaluation_prompt)
```

```python
messages = [{"role": "user", "content": evaluation_prompt}]
response = openai.chat.completions.create(model=MODEL, messages=messages)
print(response.choices[0].message.content)
```
